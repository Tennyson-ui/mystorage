```python
import importlib
import os
import platform
from pathlib import Path

print('OK Python:', platform.python_version())
print('OK diretorio atual:', Path.cwd())
print('OK HOME:', os.environ.get('HOME', '(nao definido)'))
```

```python
packages = [
    'earthaccess',
    'xarray',
    'netCDF4',
    'h5netcdf',
    'pandas',
    'geopandas',
    'shapely',
    'pyproj',
    'matplotlib',
    'ipyleaflet',
    'ipywidgets',
    'fiona',
    'pyogrio',
    'numpy',
    'requests',
    'pyarrow',
]

for package in packages:
    try:
        module = importlib.import_module(package)
        version = getattr(module, '__version__', 'versao nao informada')
        print(f'OK {package}: {version}')
    except Exception as exc:
        print(f'FALHA {package}: {exc}')
```

```python
def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / 'requirements.txt').exists() and (candidate / 'src' / 'check_environment.py').exists():
            return candidate
    fallback = Path.home() / 'mystorage' / 'PPGGAG1889' / 'atividade3_swot'
    if fallback.exists():
        return fallback
    raise RuntimeError('FALHA: abra/rode este notebook dentro do repositorio atividade3_swot ou faça cd para a raiz do projeto.')

project_root = find_project_root(Path.cwd().resolve())
output_dir = project_root / 'outputs' / 'logs'
output_dir.mkdir(parents=True, exist_ok=True)
test_file = output_dir / 'check_notebook.tmp'

try:
    test_file.write_text('ok\n', encoding='utf-8')
    test_file.unlink(missing_ok=True)
    print('OK raiz do projeto:', project_root)
    print('OK escrita em:', output_dir)
except Exception as exc:
    print('FALHA escrita:', exc)
```
